In the code below, notice how we use the derivatives we just derived to update the parameters `w` and `b` in every iteration of the loop.

```
import numpy as np

def gradient_descent(x, y, iterations=1000, learning_rate=0.01):
    w = 0.0  # Initial weight
    b = 0.0  # Initial bias
    n = len(x)

    for i in range(iterations):
        # 1. The Prediction (Current line: y = wx + b)
        y_predicted = w * x + b

        # 2. Calculate the Derivatives (The Gradient)
        # dw is the partial derivative of MSE with respect to w
        # db is the partial derivative of MSE with respect to b
        dw = (2/n) * np.sum(x * (y_predicted - y))
        db = (2/n) * np.sum(y_predicted - y)

        # 3. Update Parameters
        w = w - (learning_rate * dw)
        b = b - (learning_rate * db)

        # Optional: Print progress every 100 iterations
        if i % 100 == 0:
            cost = np.mean(np.square(y - y_predicted))
            print(f"Iteration {i}: Cost {cost:.4f}, w {w:.4f}, b {b:.4f}")

    return w, b

# Example Data
x_data = np.array([1, 2, 3, 4, 5])
y_data = np.array([5, 7, 9, 11, 13]) # y = 2x + 3

final_w, final_b = gradient_descent(x_data, y_data)
print(f"\nFinal Result: y = {final_w:.2f}x + {final_b:.2f}")
```

### Key Takeaways for Your Projects

* **Vectorization:** In the code above, `x * (y_predicted - y)` is a vectorized operation. Using NumPy makes these calculations significantly faster than using standard Python lists and loops, which is crucial when your datasets grow.
